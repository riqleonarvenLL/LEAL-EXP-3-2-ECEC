A. POSITIONAL AND LABEL-BASED SLICING

```python
# Part A: Positional and Label-Based Slicing
import pandas as pd
```

```python
# Load dataset
cars = pd.read_csv("cars.csv")
```

```python
# a. Display shape and column names
print("Shape:", cars.shape)
print("Columns:", list(cars.columns))
```

```python
# b. Rows 6 to 10 using iloc (Python is 0-indexed)
cars_6_to_10 = cars.iloc[5:10]
print(cars_6_to_10)
```

```python
# c. Select specific columns by label
cars_6_to_10_selected = cars_6_to_10[["Model", "mpg", "cyl", "hp", "gear"]]
print(cars_6_to_10_selected)
```

B. MODEL LOOKUP

```python
toyota = cars[cars["Model"] == "Toyota Corolla"]
print(toyota)
```

```python
# b. Pontiac Firebird with selected columns
pontiac = cars.loc[cars["Model"] == "Pontiac Firebird", ["Model", "mpg", "hp", "wt"]]
print(pontiac)
```

C. MULTI-MODEL SUBSETTING

```python
selected_cars = cars[cars["Model"].isin(["Datsun 710", "Lotus Europa", "Ferrari Dino"])]
```

```python
selected_cars = selected_cars[["Model", "mpg", "cyl", "hp", "gear"]]
```

```python

# Display results and shape
print(selected_cars)
print("Shape:", selected_cars.shape)
```
